```python
def show_video_grid(video_name, interval_sec=5, cols=6):
    path = os.path.join(aboda_path, video_name)
    cap = cv2.VideoCapture(path)
    fps = cap.get(cv2.CAP_PROP_FPS)
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    duration = total_frames / fps

    print(f"{video_name}: {total_frames} frames, {fps:.1f} fps, {duration:.1f} sec")

    frame_step = int(fps * interval_sec)
    frame_indices = list(range(0, total_frames, frame_step))

    rows = (len(frame_indices) + cols - 1) // cols
    fig, axes = plt.subplots(rows, cols, figsize=(cols*2.5, rows*2.5))
    axes = axes.flatten() if rows > 1 else [axes] if cols == 1 else axes

    for i, idx in enumerate(frame_indices):
        cap.set(cv2.CAP_PROP_POS_FRAMES, idx)
        ret, frame = cap.read()
        if ret:
            frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
            axes[i].imshow(frame_rgb)
            axes[i].set_title(f"f{idx}", fontsize=9)
        axes[i].axis('off')

    for j in range(len(frame_indices), len(axes)):
        axes[j].axis('off')

    plt.tight_layout()
    plt.show()
    cap.release()
    return total_frames, fps

# Try it on video1 first
show_video_grid('video1.avi', interval_sec=5)
```

```python
def zoom_range(video_name, start_frame, end_frame, step=10):
    path = os.path.join(aboda_path, video_name)
    cap = cv2.VideoCapture(path)

    frame_indices = list(range(start_frame, end_frame + 1, step))
    cols = 5
    rows = (len(frame_indices) + cols - 1) // cols
    fig, axes = plt.subplots(rows, cols, figsize=(cols*3, rows*3))
    axes = axes.flatten() if rows > 1 else axes

    for i, idx in enumerate(frame_indices):
        cap.set(cv2.CAP_PROP_POS_FRAMES, idx)
        ret, frame = cap.read()
        if ret:
            frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
            axes[i].imshow(frame_rgb)
            axes[i].set_title(f"f{idx}", fontsize=10)
        axes[i].axis('off')

    for j in range(len(frame_indices), len(axes)):
        axes[j].axis('off')

    plt.tight_layout()
    plt.show()
    cap.release()

# Zoom into the window where the bag gets left behind
zoom_range('video1.avi', 1341, 1490, step=15)
```

```python
def clean_mask(fg_mask):
    # Remove shadows (gray=127) - keep only pure foreground (white=255)
    _, mask = cv2.threshold(fg_mask, 200, 255, cv2.THRESH_BINARY)
    # Morphological opening (erode then dilate) - removes small noise specks
    kernel = np.ones((5,5), np.uint8)
    mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel)
    # Morphological closing (dilate then erode) - fills small holes in real objects
    mask = cv2.morphologyEx(mask, cv2.MORPH_CLOSE, kernel)
    return mask

cleaned = clean_mask(fg_mask)

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(fg_mask, cmap='gray')
plt.title("Raw mask")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(cleaned, cmap='gray')
plt.title("Cleaned mask")
plt.axis('off')
plt.show()
```

```python
def get_reference_background(video_path, n_frames=100):
    """Build a background model ONLY from the first N frames (before any object drop),
    then freeze it - don't let it keep adapting."""
    cap = cv2.VideoCapture(video_path)
    backSub = cv2.createBackgroundSubtractorMOG2(history=n_frames, varThreshold=16, detectShadows=False)

    for i in range(n_frames):
        ret, frame = cap.read()
        if not ret:
            break
        backSub.apply(frame)

    bg_model = backSub.getBackgroundImage()
    cap.release()
    return bg_model

# Test on video1
video_path = os.path.join(aboda_path, 'video1.avi')
reference_bg = get_reference_background(video_path, n_frames=100)

plt.figure(figsize=(6,4))
plt.imshow(cv2.cvtColor(reference_bg, cv2.COLOR_BGR2RGB))
plt.title("Frozen reference background")
plt.axis('off')
plt.show()
```

```python
def get_frame_at(video_path, frame_idx):
    cap = cv2.VideoCapture(video_path)
    cap.set(cv2.CAP_PROP_POS_FRAMES, frame_idx)
    ret, frame = cap.read()
    cap.release()
    return frame

def diff_against_reference(frame, reference_bg, thresh=30):
    gray_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    gray_ref = cv2.cvtColor(reference_bg, cv2.COLOR_BGR2GRAY)
    diff = cv2.absdiff(gray_frame, gray_ref)
    _, mask = cv2.threshold(diff, thresh, 255, cv2.THRESH_BINARY)
    kernel = np.ones((5,5), np.uint8)
    mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel)
    mask = cv2.morphologyEx(mask, cv2.MORPH_CLOSE, kernel)
    return mask

frame_1450 = get_frame_at(video_path, 1450)
diff_mask = diff_against_reference(frame_1450, reference_bg)

plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(frame_1450, cv2.COLOR_BGR2RGB))
plt.title("Frame 1450")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(diff_mask, cmap='gray')
plt.title("Diff vs frozen background")
plt.axis('off')
plt.show()
```

```python
def get_blobs(mask, min_area=200, max_area=15000):
    contours, _ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)
    blobs = []
    for c in contours:
        area = cv2.contourArea(c)
        if min_area < area < max_area:
            x, y, w, h = cv2.boundingRect(c)
            blobs.append({'bbox': (x, y, w, h), 'area': area, 'aspect_ratio': w/h})
    return blobs

blobs = get_blobs(diff_mask)
print(f"Found {len(blobs)} candidate blobs:")
for b in blobs:
    print(b)

# Visualize
frame_viz = frame_1450.copy()
for b in blobs:
    x, y, w, h = b['bbox']
    cv2.rectangle(frame_viz, (x, y), (x+w, y+h), (0, 255, 0), 2)

plt.figure(figsize=(8,5))
plt.imshow(cv2.cvtColor(frame_viz, cv2.COLOR_BGR2RGB))
plt.title("Detected blobs (candidate objects)")
plt.axis('off')
plt.show()
```

```python
from ultralytics import YOLO

yolo_model = YOLO('yolov8n.pt')  # downloads automatically, nano = fastest

def detect_people(frame):
    results = yolo_model(frame, classes=[0], verbose=False)  # class 0 = person in COCO
    people_boxes = []
    for r in results:
        for box in r.boxes:
            x1, y1, x2, y2 = box.xyxy[0].cpu().numpy()
            people_boxes.append((int(x1), int(y1), int(x2), int(y2)))
    return people_boxes

def boxes_overlap(box1, box2, pad=20):
    # box1: (x,y,w,h) blob format, box2: (x1,y1,x2,y2) YOLO format
    x, y, w, h = box1
    bx1, by1, bx2, by2 = x-pad, y-pad, x+w+pad, y+h+pad
    px1, py1, px2, py2 = box2
    return not (bx2 < px1 or px2 < bx1 or by2 < py1 or py2 < by1)

people = detect_people(frame_1450)
print(f"Detected {len(people)} people:", people)

object_candidates = []
for b in blobs:
    overlaps_person = any(boxes_overlap(b['bbox'], p) for p in people)
    if not overlaps_person:
        object_candidates.append(b)

print(f"\nObject candidates (not overlapping any person): {len(object_candidates)}")
for b in object_candidates:
    print(b)

# Visualize: people in red, object candidates in green
frame_viz2 = frame_1450.copy()
for p in people:
    cv2.rectangle(frame_viz2, (p[0],p[1]), (p[2],p[3]), (0,0,255), 2)
for b in object_candidates:
    x, y, w, h = b['bbox']
    cv2.rectangle(frame_viz2, (x,y), (x+w,y+h), (0,255,0), 2)

plt.figure(figsize=(8,5))
plt.imshow(cv2.cvtColor(frame_viz2, cv2.COLOR_BGR2RGB))
plt.title("Red=people, Green=unattended object candidates")
plt.axis('off')
plt.show()
```

```python
def process_video_for_abandonment(video_path, bg_warmup_frames=100,
                                     check_every=5, persistence_frames=90,
                                     position_tolerance=30):
    """
    check_every: only run detection every N frames (speed - don't need every single frame)
    persistence_frames: how many frames (in real video time) an object must stay
                         unattended at ~same spot before we call it abandoned
    position_tolerance: pixel distance to consider 'same location' between checks
    """
    reference_bg = get_reference_background(video_path, n_frames=bg_warmup_frames)
    cap = cv2.VideoCapture(video_path)
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

    candidate_tracker = {}  # {approx_position: {'first_seen': frame_idx, 'last_seen': frame_idx}}
    abandonment_frame = None

    frame_idx = bg_warmup_frames
    cap.set(cv2.CAP_PROP_POS_FRAMES, frame_idx)

    while frame_idx < total_frames:
        ret, frame = cap.read()
        if not ret:
            break

        if frame_idx % check_every == 0:
            mask = diff_against_reference(frame, reference_bg)
            blobs = get_blobs(mask)
            people = detect_people(frame)
            candidates = [b for b in blobs if not any(boxes_overlap(b['bbox'], p) for p in people)]

            current_positions = []
            for c in candidates:
                x, y, w, h = c['bbox']
                center = (x + w//2, y + h//2)
                current_positions.append(center)

                matched_key = None
                for key in candidate_tracker:
                    if abs(key[0]-center[0]) < position_tolerance and abs(key[1]-center[1]) < position_tolerance:
                        matched_key = key
                        break

                if matched_key:
                    candidate_tracker[matched_key]['last_seen'] = frame_idx
                    duration = candidate_tracker[matched_key]['last_seen'] - candidate_tracker[matched_key]['first_seen']
                    if duration >= persistence_frames and abandonment_frame is None:
                        abandonment_frame = candidate_tracker[matched_key]['first_seen']
                        cap.release()
                        return abandonment_frame, candidate_tracker
                else:
                    candidate_tracker[center] = {'first_seen': frame_idx, 'last_seen': frame_idx}

            # Prune trackers not seen recently (object picked back up / was noise)
            stale_keys = [k for k,v in candidate_tracker.items()
                          if frame_idx - v['last_seen'] > persistence_frames and k not in current_positions]
            for k in stale_keys:
                del candidate_tracker[k]

        frame_idx += 1

    cap.release()
    return abandonment_frame, candidate_tracker

# Test on video1 - we expect something near our labeled frame 1431
detected_frame, tracker = process_video_for_abandonment(
    os.path.join(aboda_path, 'video1.avi'),
    persistence_frames=90  # 3 seconds at 30fps
)
print(f"Detected abandonment at frame: {detected_frame}")
print(f"Ground truth label: 1431")
```

```python
import time

results_summary = []

for video_name, label_info in aboda_labels.items():
    video_path = os.path.join(aboda_path, video_name)
    gt_frame = label_info['abandonment_frame']

    print(f"Processing {video_name}...")
    start = time.time()

    try:
        detected_frame, _ = process_video_for_abandonment(video_path, persistence_frames=90)
    except Exception as e:
        print(f"  Error: {e}")
        detected_frame = None

    elapsed = time.time() - start

    if gt_frame is None:
        # Negative case - we want detected_frame to also be None (no false alarm)
        correct = (detected_frame is None)
        error = None
        result_type = "TN (correct no-alarm)" if correct else "FP (false alarm!)"
    else:
        if detected_frame is None:
            correct = False
            error = None
            result_type = "FN (missed detection)"
        else:
            error = abs(detected_frame - gt_frame)
            correct = error <= 90  # within 3 seconds tolerance
            result_type = "TP (correct)" if correct else "TP but late/early"

    results_summary.append({
        'video': video_name,
        'ground_truth': gt_frame,
        'detected': detected_frame,
        'error_frames': error,
        'result': result_type,
        'time_sec': round(elapsed, 1)
    })
    print(f"  GT={gt_frame}, Detected={detected_frame}, Result={result_type}, Time={elapsed:.1f}s")

print("\n\n=== SUMMARY ===")
for r in results_summary:
    print(r)
```

```python
def get_median_background(video_path, n_samples=50):
    """Sample frames spread across the whole video, take per-pixel median.
    Robust to people being present in any single frame."""
    cap = cv2.VideoCapture(video_path)
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    sample_indices = np.linspace(0, total_frames - 1, n_samples, dtype=int)

    frames = []
    for idx in sample_indices:
        cap.set(cv2.CAP_PROP_POS_FRAMES, idx)
        ret, frame = cap.read()
        if ret:
            frames.append(frame)
    cap.release()

    median_bg = np.median(np.array(frames), axis=0).astype(np.uint8)
    return median_bg

# Re-test on video6
median_bg_v6 = get_median_background(os.path.join(aboda_path, 'video6.avi'), n_samples=50)

plt.figure(figsize=(6,4))
plt.imshow(cv2.cvtColor(median_bg_v6, cv2.COLOR_BGR2RGB))
plt.title("Median background (50 samples across full video)")
plt.axis('off')
plt.show()
```

```python
def diff_against_reference_v2(frame, reference_bg, thresh=30):
    gray_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    gray_ref = cv2.cvtColor(reference_bg, cv2.COLOR_BGR2GRAY)
    diff = cv2.absdiff(gray_frame, gray_ref)
    _, mask = cv2.threshold(diff, thresh, 255, cv2.THRESH_BINARY)
    kernel = np.ones((5,5), np.uint8)
    mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel)
    mask = cv2.morphologyEx(mask, cv2.MORPH_CLOSE, kernel)
    return mask

def process_video_for_abandonment_v2(video_path, n_bg_samples=50,
                                       check_every=5, persistence_frames=90,
                                       position_tolerance=30):
    reference_bg = get_median_background(video_path, n_samples=n_bg_samples)
    cap = cv2.VideoCapture(video_path)
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

    candidate_tracker = {}
    abandonment_frame = None

    frame_idx = 0
    cap.set(cv2.CAP_PROP_POS_FRAMES, frame_idx)

    while frame_idx < total_frames:
        ret, frame = cap.read()
        if not ret:
            break

        if frame_idx % check_every == 0:
            mask = diff_against_reference_v2(frame, reference_bg)
            blobs = get_blobs(mask)
            people = detect_people(frame)
            candidates = [b for b in blobs if not any(boxes_overlap(b['bbox'], p) for p in people)]

            current_positions = []
            for c in candidates:
                x, y, w, h = c['bbox']
                center = (x + w//2, y + h//2)
                current_positions.append(center)

                matched_key = None
                for key in candidate_tracker:
                    if abs(key[0]-center[0]) < position_tolerance and abs(key[1]-center[1]) < position_tolerance:
                        matched_key = key
                        break

                if matched_key:
                    candidate_tracker[matched_key]['last_seen'] = frame_idx
                    duration = candidate_tracker[matched_key]['last_seen'] - candidate_tracker[matched_key]['first_seen']
                    if duration >= persistence_frames and abandonment_frame is None:
                        abandonment_frame = candidate_tracker[matched_key]['first_seen']
                        cap.release()
                        return abandonment_frame, candidate_tracker
                else:
                    candidate_tracker[center] = {'first_seen': frame_idx, 'last_seen': frame_idx}

            stale_keys = [k for k,v in candidate_tracker.items()
                          if frame_idx - v['last_seen'] > persistence_frames and k not in current_positions]
            for k in stale_keys:
                del candidate_tracker[k]

        frame_idx += 1

    cap.release()
    return abandonment_frame, candidate_tracker

results_summary_v2 = []

for video_name, label_info in aboda_labels.items():
    video_path = os.path.join(aboda_path, video_name)
    gt_frame = label_info['abandonment_frame']

    print(f"Processing {video_name}...")
    start = time.time()
    detected_frame, _ = process_video_for_abandonment_v2(video_path, persistence_frames=90)
    elapsed = time.time() - start

    if gt_frame is None:
        correct = (detected_frame is None)
        error = None
        result_type = "TN (correct no-alarm)" if correct else "FP (false alarm!)"
    else:
        if detected_frame is None:
            correct = False
            error = None
            result_type = "FN (missed detection)"
        else:
            error = abs(detected_frame - gt_frame)
            correct = error <= 90
            result_type = "TP (correct)" if correct else "TP but late/early"

    results_summary_v2.append({
        'video': video_name, 'ground_truth': gt_frame, 'detected': detected_frame,
        'error_frames': error, 'result': result_type, 'time_sec': round(elapsed, 1)
    })
    print(f"  GT={gt_frame}, Detected={detected_frame}, Result={result_type}, Time={elapsed:.1f}s")
```

```python
def diff_against_reference_v3(frame, reference_bg, thresh=30):
    gray_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY).astype(float)
    gray_ref = cv2.cvtColor(reference_bg, cv2.COLOR_BGR2GRAY).astype(float)

    # Match frame's overall brightness to reference's before comparing
    scale = (gray_ref.mean() + 1e-6) / (gray_frame.mean() + 1e-6)
    gray_frame_norm = np.clip(gray_frame * scale, 0, 255).astype(np.uint8)
    gray_ref_uint8 = gray_ref.astype(np.uint8)

    diff = cv2.absdiff(gray_frame_norm, gray_ref_uint8)
    _, mask = cv2.threshold(diff, thresh, 255, cv2.THRESH_BINARY)
    kernel = np.ones((5,5), np.uint8)
    mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel)
    mask = cv2.morphologyEx(mask, cv2.MORPH_CLOSE, kernel)
    return mask

# Re-test on video7 frame 0
mask_0_v7_fixed = diff_against_reference_v3(frame_0_v7, median_bg_v7)

plt.figure(figsize=(8,5))
plt.imshow(mask_0_v7_fixed, cmap='gray')
plt.title("Diff mask at frame 0 (brightness-normalized)")
plt.axis('off')
plt.show()

blobs_fixed = get_blobs(mask_0_v7_fixed)
print(f"Blobs found: {len(blobs_fixed)}")
for b in blobs_fixed:
    print(b)
```

```python
def get_smart_reference_background(video_path, target_clean_frames=100, max_scan=1000):
    """Scan forward through the video, feeding only people-free frames
    into the background model, until we've got enough clean frames."""
    cap = cv2.VideoCapture(video_path)
    backSub = cv2.createBackgroundSubtractorMOG2(history=target_clean_frames, varThreshold=16, detectShadows=False)

    clean_count = 0
    scanned = 0
    while clean_count < target_clean_frames and scanned < max_scan:
        ret, frame = cap.read()
        if not ret:
            break
        scanned += 1
        people = detect_people(frame)
        if len(people) == 0:
            backSub.apply(frame)
            clean_count += 1

    bg_model = backSub.getBackgroundImage()
    cap.release()
    return bg_model, scanned

# Test on video6 (the one with a person present at the very start)
smart_bg_v6, scanned = get_smart_reference_background(os.path.join(aboda_path, 'video6.avi'))
print(f"Scanned {scanned} frames to get 100 clean ones")

plt.figure(figsize=(6,4))
plt.imshow(cv2.cvtColor(smart_bg_v6, cv2.COLOR_BGR2RGB))
plt.title("Smart (people-free) reference background")
plt.axis('off')
plt.show()
```

```python
def process_video_final(video_path, target_clean_frames=100, check_every=5,
                          persistence_frames=90, position_tolerance=30, learning_rate=0.0005):
    cap = cv2.VideoCapture(video_path)
    backSub = cv2.createBackgroundSubtractorMOG2(history=500, varThreshold=16, detectShadows=False)

    # Smart warmup: only learn from people-free frames
    clean_count = 0
    frame_idx = 0
    while clean_count < target_clean_frames and frame_idx < 1000:
        ret, frame = cap.read()
        if not ret:
            break
        frame_idx += 1
        people = detect_people(frame)
        if len(people) == 0:
            backSub.apply(frame, learningRate=-1)  # normal learning during warmup
            clean_count += 1

    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    candidate_tracker = {}
    abandonment_frame = None

    while frame_idx < total_frames:
        ret, frame = cap.read()
        if not ret:
            break

        # Continue updating background VERY slowly (tracks lighting drift,
        # but too slow to absorb a new object within our persistence window)
        fg_mask = backSub.apply(frame, learningRate=learning_rate)

        if frame_idx % check_every == 0:
            _, mask = cv2.threshold(fg_mask, 200, 255, cv2.THRESH_BINARY)
            kernel = np.ones((5,5), np.uint8)
            mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, kernel)
            mask = cv2.morphologyEx(mask, cv2.MORPH_CLOSE, kernel)

            blobs = get_blobs(mask)
            people = detect_people(frame)
            candidates = [b for b in blobs if not any(boxes_overlap(b['bbox'], p) for p in people)]

            current_positions = []
            for c in candidates:
                x, y, w, h = c['bbox']
                center = (x + w//2, y + h//2)
                current_positions.append(center)

                matched_key = None
                for key in candidate_tracker:
                    if abs(key[0]-center[0]) < position_tolerance and abs(key[1]-center[1]) < position_tolerance:
                        matched_key = key
                        break

                if matched_key:
                    candidate_tracker[matched_key]['last_seen'] = frame_idx
                    duration = candidate_tracker[matched_key]['last_seen'] - candidate_tracker[matched_key]['first_seen']
                    if duration >= persistence_frames and abandonment_frame is None:
                        abandonment_frame = candidate_tracker[matched_key]['first_seen']
                        cap.release()
                        return abandonment_frame
                else:
                    candidate_tracker[center] = {'first_seen': frame_idx, 'last_seen': frame_idx}

            stale_keys = [k for k,v in candidate_tracker.items()
                          if frame_idx - v['last_seen'] > persistence_frames and k not in current_positions]
            for k in stale_keys:
                del candidate_tracker[k]

        frame_idx += 1

    cap.release()
    return abandonment_frame

# Quick test on video7 and video6 first before running the full suite
for vname in ['video6.avi', 'video7.avi']:
    gt = aboda_labels[vname]['abandonment_frame']
    start = time.time()
    det = process_video_final(os.path.join(aboda_path, vname))
    print(f"{vname}: GT={gt}, Detected={det}, Time={time.time()-start:.1f}s")
```

```python
# Re-run the Step 16h version (standard adaptive MOG2, 100-frame warmup) for clean final numbers
final_results = []

for video_name, label_info in aboda_labels.items():
    video_path = os.path.join(aboda_path, video_name)
    gt_frame = label_info['abandonment_frame']

    detected_frame, _ = process_video_for_abandonment(video_path, persistence_frames=90)

    if gt_frame is None:
        correct = (detected_frame is None)
        error = None
        category = "TN" if correct else "FP"
    else:
        if detected_frame is None:
            correct = False
            error = None
            category = "FN"
        else:
            error = abs(detected_frame - gt_frame)
            correct = error <= 150  # 5 second tolerance at 30fps
            category = "TP" if correct else "TP (late)"

    final_results.append({
        'video': video_name, 'ground_truth': gt_frame, 'detected': detected_frame,
        'error_frames': error, 'error_sec': round(error/30, 1) if error else None,
        'category': category
    })

# Compute summary metrics
tp = sum(1 for r in final_results if r['category'] == 'TP')
tp_late = sum(1 for r in final_results if r['category'] == 'TP (late)')
fn = sum(1 for r in final_results if r['category'] == 'FN')
fp = sum(1 for r in final_results if r['category'] == 'FP')
tn = sum(1 for r in final_results if r['category'] == 'TN')

positives = tp + tp_late + fn  # videos with actual ground truth
detected_at_all = tp + tp_late  # detected something, regardless of timing accuracy

detection_rate = detected_at_all / positives if positives else 0
false_alarm_rate = fp / (fp + tn) if (fp + tn) else 0
within_tolerance_rate = tp / positives if positives else 0

errors = [r['error_frames'] for r in final_results if r['error_frames'] is not None]
mean_error_sec = (sum(errors)/len(errors))/30 if errors else None

print("=== FINAL BASELINE RESULTS ===\n")
for r in final_results:
    print(r)

print(f"\n=== METRICS ===")
print(f"Detection rate (found event at all): {detected_at_all}/{positives} = {detection_rate:.1%}")
print(f"Within 5-sec tolerance: {tp}/{positives} = {within_tolerance_rate:.1%}")
print(f"False alarm rate (on negatives): {fp}/{fp+tn} = {false_alarm_rate:.1%}")
print(f"Mean timing error (all matched): {mean_error_sec:.1f} sec" if mean_error_sec else "N/A")

# Save results
with open('/content/drive/MyDrive/J26-DS-320/datasets/abandoned_object_baseline_results.json', 'w') as f:
    json.dump({
        'results': final_results,
        'metrics': {
            'detection_rate': detection_rate,
            'within_tolerance_rate': within_tolerance_rate,
            'false_alarm_rate': false_alarm_rate,
            'mean_error_sec': mean_error_sec
        }
    }, f, indent=2)
print("\nSaved to Drive.")
```
